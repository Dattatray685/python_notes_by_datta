```python
heights = [160,140,150,180,160,170]
weights = [65,70,68,64,72,90]
#plt a line plot based on height and weight and check realtionship between height and weight
#plt.plot(x,y)
plt.plot(weights,heights)
plt.show()
#Note:- In matplotlib the default plot is a line plot
```

```python
#instead of line plot plot it with scatters or any markers
plt.plot(weights,heights,"o:g")  #o,*,x,P,p,H,h,+
plt.show()
#4 line type formats
#1. solid line   (-)
#2. Dashed line  (--)
#3. Dotted line  (:)
#4. Dash dot line(-.)
```

```python
#change the color of scatter and scatter edge color --> mec and mfc
plt.plot(weights,heights,"o:g",mec="#fc6f03",mfc="#9003fc",ms=10)  #o,*,x,P,p,H,h,+
plt.show()
```

```python
#syntax :-  plt.plot(x,y,"marke:color",mec=,mfc=,ms)
#check the relaintion betweeon two features we use line plot (ex:- revenue generate every year )
```

**scatter plot**

```python
#plt.scatter(x,y,color=,s=)
plt.scatter(weights,heights,color="#9003fc",s=20)
plt.show()
#note: scatter is used to check relation between two numerical features (continues data)
```

**Histogram**

```python
#Histogram can be drawn based on frequency
#In histogram the y-axis is nothing but frequency
#Histogram used to check distribution values in your data
data = [60,71,80,90,54,33,65,56,76,89,56,88,43,42]
import matplotlib.pyplot as plt
plt.hist(data,bins=20,color="#fcba03")
plt.show()
#syntax:- plt.hist(x,bins=,color=)
```

**Bar Plot**

```python
names = ["A","B","C","D"]
values = [10,25,15,30]
#check the relation between categorical data and numerical data
#syntax: plt.bar(x,y,bins,color)
plt.bar(names,values)
plt.xlabel("Names")
plt.ylabel("Marks out of 50")
plt.show()
#comparing catgeories  w.r.t numerical values showing count,total
```

**PiePlot** --> is drwan based on frequency

```python
plt.pie(values,labels=names,autopct="%1.1f%%")
plt.xlabel("Names")
plt.ylabel("Marks out of 50")
plt.show()
```

```python
import seaborn as sns
```

```python
#load the dataset of flights
df = sns.load_dataset("flights")
df.head()
#how did the number of passengers changes over time
plt.plot(df['year'],df['passengers'])
plt.title("passengers vs Year")
plt.xlabel("Year")
plt.ylabel("Passengers")
plt.show()
```

```python
#apply sum() aggregate
year_data = df.groupby("year")['passengers'].sum()
year_data
```

```python
plt.plot(year_data.index,year_data.values)
plt.title("passengers vs Year")
plt.xlabel("Year")
plt.ylabel("Passengers")
plt.show()
```

```python
#sns.linplot(data=,x=,y=)
sns.lineplot(data=df,x="year",y="passengers")
plt.title("passengers vs Year")
plt.xlabel("Year")
plt.ylabel("Passengers")
plt.show()
```

**BarPlot***

```python
#what is the average passengers permonth?
month_data = df.groupby("month")['passengers'].mean()
month_data
```

```python
plt.bar(month_data.index,month_data.values)
plt.title("passengers vs Year")
plt.xlabel("Year")
plt.ylabel("Passengers")
plt.show()
```

```python
sns.barplot(data=df,x="month",y="passengers")
plt.xticks(rotation=45)
plt.title("passengers vs Year")
plt.xlabel("Year")
plt.ylabel("Passengers")
plt.show()
```

```python
#what percentage contributes per each month in overall data
month_data = df.groupby("month")['passengers'].sum()
plt.pie(month_data,labels=month_data.index,autopct="%1.1f%%")
plt.title("passengers vs Year")
plt.xlabel("Year")
plt.ylabel("Passengers")
plt.show()
```

```python
sns.pieplot(data=df,x="passengers",labels="month")
#there is no pieplot in seaborn
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load the data "tips"
df = sns.load_dataset("tips")
df.head()
```

```python
#to check the distribution 2 continuous variables use the scatterplot
#plt.scatter(x,y,color=,s)
#check the distribution and relation between total_bill and tip
plt.scatter(df['tip'],df['total_bill'])
plt.xlabel("tip")
plt.ylabel("total_bill")
plt.show()
```

```python
#syntax:  sns.scatterplot(data=,x=,y=,hue=,size=,style)
#hue assigns different colors based category
sns.scatterplot(data=df,x="tip",y="total_bill",hue="sex",style="smoker")
plt.xlabel("tip")
plt.ylabel("total_bill")
plt.show()
```

```python
#draw plot to check on which more bills can be paid
sns.scatterplot(data=df,x="tip",y="total_bill",hue="day")
plt.xlabel("tip")
plt.ylabel("total_bill")
plt.show()
```

**BoxPlot**

```python
#boxplot is used to check the outliers
#what is outliers?
#extreme range of values
#[1,2,3,4,5,6,7,8,9,1000,10000]
#check the outliers present in tips data
#syntax:   sns.boxplot(data=,x=,y=,hue=,color=)
sns.boxplot(data=df,x="sex",y="total_bill",hue='day')
plt.xlabel("tip")
plt.ylabel("total_bill")
plt.show()
```

```python
#what are all components do we have in a boxplot?
#1.lower limit (min) --> 5% of data
#2.Q1 --> quartile --> 25% of data
#3.Q2 --> Median --> 50%  of data
#4.Q3 --> 75% of data
#5.upper limit(max) --> 95% of data
```

```python
plt.boxplot(df['total_bill'])
```

**Violinplot**

```python
sns.violinplot(data=df,x="sex",y="total_bill",hue='day')
plt.xlabel("tip")
plt.ylabel("total_bill")
plt.show()
```

```python
#line plot :-     sns.lineplot(data=,x=,y=,hue=,style=,size=)
#histogram :-     sns.histplot(data=,x=,hue=,bins=)
#scatterplot :-   sns.scatterplot(data=,x=,y=,hue=,style=,size=)
#barplot :-       sns.barplot(data=,x=,y=,bins=,hue=)
#boxplot:-        sns.boxplot(data=,x=,y=,hue=)
```

**statistical estimation plot**

```python
#sns.lmplot(data=,x=,y=,hue)
sns.lmplot(data=df,x="tip",y="total_bill")
```
